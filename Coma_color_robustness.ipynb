```python
from collections import OrderedDict
import math
import numpy as np
import pandas as pd


def fmt_pval(p):
    """Format p-values for LaTeX."""
    if pd.isna(p):
        return "--"
    if p < 1e-3:
        s = f"{p:.2e}"
        mant, exp = s.split("e")
        exp = int(exp)
        return rf"${mant}\times10^{{{exp}}}$"
    return f"{p:.3f}"


def fmt_num(x, ndp=3, signed=False):
    """Format a float for LaTeX table output."""
    if pd.isna(x):
        return "--"
    if signed:
        return f"{x:+.{ndp}f}"
    return f"{x:.{ndp}f}"


def fmt_slope_sci(row, ndp=2):
    """
    Format slope and asymmetric CI as:
    $m_{-lo}^{+hi}\\times10^{exp}$

    Example:
    -0.16346 with CI [-0.19645, -0.13168]
    -> $-1.63_{-0.33}^{+0.32}\\times10^{-1}$
    """
    slope = row.get("slope", np.nan)
    lo = row.get("slope_lo", np.nan)
    hi = row.get("slope_hi", np.nan)

    if any(pd.isna(v) for v in [slope, lo, hi]):
        return "--"

    if slope == 0:
        return rf"${0:.{ndp}f}$"

    exp = int(math.floor(math.log10(abs(slope))))
    scale = 10 ** exp

    m = slope / scale
    dm_lo = (slope - lo) / scale
    dm_hi = (hi - slope) / scale

    return rf"${m:.{ndp}f}_{{-{dm_lo:.{ndp}f}}}^{{+{dm_hi:.{ndp}f}}}\times10^{{{exp}}}$"


def pretty_metric_name(name):
    """Map internal metric names to nicer LaTeX labels."""
    mapping = {
        "R_bcg_mid_arcmin": "BCG midpoint",
        "gc_bg_voronoi_density_arcsec2_inv": r"Local GC density",
        "log_gc_bg_voronoi_density": r"Local GC density",
        "log_tidal_proxy": r"BCG tidal proxy",
        "R_nearBCG_arcmin": r"BCG proximity",
    }
    return mapping.get(name, name.replace("_", r"\_"))


def pretty_sample_name(name):
    """Map sample_definition values to nicer labels."""
    mapping = {
        "fiducial": "Fiducial",
        "high_support": "High-support",
        "bright_cut": "Bright cut",
        "small_ap": r"Smaller aperture",
        "large_ap": r"Larger aperture",
    }
    return mapping.get(name, name.replace("_", r"\_"))


def robust_rows_to_deluxetable(
    robust_rows,
    caption="Robustness of the environmental trends in $\\Delta\\langle P_{\\rm blue}\\rangle_{\\rm bgsub}$.",
    label="tab:env_robustness_summary",
    include_metrics=None,
    include_samples=None,
    drop_linear_density_if_log_present=True,
    sort_samples=None,
    sort_metrics=None,
    tablecomments=None,
    ndp_rho=3,
    ndp_delta=3,
    ndp_slope=3,
):
    """
    Convert a list of dicts 'robust_rows' into an AASTeX deluxetable string.

    Expected dict keys:
      - test_name
      - sample_definition
      - N
      - rho
      - p
      - slope
      - slope_lo
      - slope_hi
      - delta_inner_outer
    """
    df = pd.DataFrame(robust_rows).copy()

    required = ["test_name", "sample_definition", "N", "rho", "p", "slope", "slope_lo", "slope_hi", "delta_inner_outer"]
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns in robust_rows: {missing}")

    # Optional filtering
    if include_samples is not None:
        df = df[df["sample_definition"].isin(include_samples)].copy()

    if include_metrics is not None:
        df = df[df["test_name"].isin(include_metrics)].copy()

    # Drop linear density if log-density exists for same sample and that option is requested
    if drop_linear_density_if_log_present:
        keep_rows = []
        grouped = df.groupby("sample_definition", sort=False)
        for sample, g in grouped:
            names = set(g["test_name"])
            for _, row in g.iterrows():
                if (
                    row["test_name"] == "gc_bg_voronoi_density_arcsec2_inv"
                    and "log_gc_bg_voronoi_density" in names
                ):
                    continue
                keep_rows.append(row)
        df = pd.DataFrame(keep_rows)

    # Sorting
    if sort_samples is None:
        sort_samples = [
            "fiducial",
            "high_support",
            "bright_cut",
            "small_ap",
            "large_ap",
        ]
    if sort_metrics is None:
        sort_metrics = [
            "R_bcg_mid_arcmin",
            "log_gc_bg_voronoi_density",
            "gc_bg_voronoi_density_arcsec2_inv",
            "log_tidal_proxy",
            "R_nearBCG_arcmin",
        ]

    sample_order = {k: i for i, k in enumerate(sort_samples)}
    metric_order = {k: i for i, k in enumerate(sort_metrics)}

    df["sample_order"] = df["sample_definition"].map(lambda x: sample_order.get(x, 999))
    df["metric_order"] = df["test_name"].map(lambda x: metric_order.get(x, 999))
    df = df.sort_values(["metric_order", "sample_order"]).reset_index(drop=True)

    # Build rows
    body_lines = []
    current_metric = None
    current_sample = None

    for _, row in df.iterrows():
        sample = row["sample_definition"]
        metric = row["test_name"]

        metric_tex = pretty_metric_name(metric) if metric != current_metric else ""
        sample_tex = pretty_sample_name(sample) if sample != current_sample else ""
        N_tex = str(int(row["N"])) if pd.notna(row["N"]) else "--"
        rho_tex = fmt_num(row["rho"], ndp=ndp_rho, signed=True)
        p_tex = fmt_pval(row["p"])
        slope_tex = fmt_slope_sci(row, ndp=ndp_slope)
        delta_tex = fmt_num(row["delta_inner_outer"], ndp=ndp_delta, signed=True)

        line = (
            f"{metric_tex} & {sample_tex} & {N_tex} & "
            f"${rho_tex}$ & {p_tex} & {slope_tex} & ${delta_tex}$ \\\\"
        )
        body_lines.append(line)

        current_metric = metric
        current_sample = sample

    if tablecomments is None:
        tablecomments = (
            "Columns give the environmental metric, sample definition, number of galaxies, "
            "Spearman rank coefficient $\\rho$, corresponding $p$-value, bootstrap slope with "
            "68\\% confidence interval, and the difference in mean residual between the inner and outer halves "
            "of the environmental variable, $\\Delta_{\\rm inner-outer}$."
        )

    table = []
    table.append(r"\begin{deluxetable*}{llccccc}")
    table.append(rf"\tablecaption{{{caption}\label{{{label}}}}}")
    table.append(r"\tablehead{")
    table.append(
        r"\colhead{Metric} & \colhead{Sample} & \colhead{$N$} & "
        r"\colhead{$\rho$} & \colhead{$p$} & \colhead{Slope} & "
        r"\colhead{$\Delta_{\rm inner-outer}$}"
    )
    table.append(r"}")
    table.append(r"\startdata")
    table.extend(body_lines)
    table.append(r"\enddata")
    table.append(rf"\tablecomments{{{tablecomments}}}")
    table.append(r"\end{deluxetable*}")

    return "\n".join(table)
```
