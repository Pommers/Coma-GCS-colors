```python
import fitz  # PyMuPDF

basepath = "/Users/richard/Documents/Astronomy/PhD/Research/Postdoc/Pasadena/"

pdf_path = basepath + "RTPomeroy_roman2026.pdf"
out_path = basepath + "roman_poster_150dpi.png"

doc = fitz.open(pdf_path)
page = doc[0]

dpi = 150
zoom = dpi / 72
mat = fitz.Matrix(zoom, zoom)

pix = page.get_pixmap(matrix=mat, alpha=False)
pix.save(out_path)

print(pix.width, pix.height)
```
