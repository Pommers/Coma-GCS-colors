```python
from astroquery.ipac.ned import Ned
import pandas as pd

galaxy_list = [
    "IC 3973", "IC 3976", "IC 3998", "IC 4011", "IC 4012", "IC 4021", "IC 4026",
    "IC 4030", "IC 4033", "IC 4040", "IC 4041", "IC 4045", "IC 4051",
    "NGC 4867", "NGC 4869", "NGC 4871", "NGC 4872", "NGC 4873", "NGC 4874",
    "NGC 4875", "NGC 4876", "NGC 4882", "NGC 4883", "NGC 4889", "NGC 4894",
    "NGC 4898", "NGC 4906", "NGC 4908"
]

results = []

for galaxy in galaxy_list:
    redshift = None
    distance = None
    dist_method = None
    try:
        obj = Ned.query_object(galaxy)
        redshift = obj['Redshift'][0]
        ra = obj['RA'].data[0]
        dec = obj['DEC'].data[0]
    except Exception:
        pass
    try:
        dist_table = Ned.get_table(galaxy, table='distances')
        if len(dist_table) > 0:
            distance = dist_table['Distance (Mpc)'][0]
            dist_method = dist_table['Method'][0]
    except Exception:
        pass
    results.append({
        'Galaxy': galaxy,
        'ra': ra,
        'dec': dec,
        'Redshift': redshift,
        'Distance (Mpc)': distance,
        'Distance Method': dist_method
    })

df = pd.DataFrame(results)
print(df)
```
